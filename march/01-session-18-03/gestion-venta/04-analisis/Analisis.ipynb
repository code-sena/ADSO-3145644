```python
import pandas as pd
import matplotlib.pyplot as plt

# Leer el archivo Excel
df = pd.read_excel('Encuesta_Respuestas.xlsx')

# Mostrar las primeras filas del DataFrame para entender su estructura
print(df.head())

# Realizar la tabulación de los datos
# Suponiendo que queremos analizar la columna 'Frecuencia_Registro_Ventas' y 'Importancia_Reportes_Ventas'
tabulacion = df.groupby(['Frecuencia_Registro_Ventas', 'Importancia_Reportes_Ventas']).size().unstack(fill_value=0)

# Mostrar la tabulación
print(tabulacion)

# Generar la primera gráfica: Gráfico de barras para cada frecuencia de registro de ventas
tabulacion.plot(kind='bar', stacked=True, figsize=(10, 7))
plt.title('Distribución de Importancia de Reportes de Ventas por Frecuencia de Registro')
plt.xlabel('Frecuencia de Registro de Ventas')
plt.ylabel('Número de Respuestas')
plt.legend(title='Importancia de Reportes de Ventas')
plt.show()

# Generar la segunda gráfica: Gráfico de pastel para una frecuencia específica
# Suponiendo que queremos analizar la primera frecuencia de registro de ventas
frecuencia_especifica = tabulacion.iloc[0]
frecuencia_especifica.plot(kind='pie', autopct='%1.1f%%', figsize=(7, 7))
plt.title(f'Distribución de Importancia de Reportes de Ventas para la Frecuencia: {tabulacion.index[0]}')
plt.ylabel('')
plt.show()

# Generar la tercera gráfica: Gráfico de barras horizontales para la importancia de reportes de ventas
tabulacion.sum().plot(kind='barh', figsize=(10, 7))
plt.title('Total de Importancia de Reportes de Ventas')
plt.xlabel('Número de Respuestas')
plt.ylabel('Importancia de Reportes de Ventas')
plt.show()

# Generar la cuarta gráfica: Gráfico de líneas para la evolución de la importancia de reportes de ventas
tabulacion.plot(kind='line', figsize=(10, 7), marker='o')
plt.title('Evolución de la Importancia de Reportes de Ventas por Frecuencia de Registro')
plt.xlabel('Frecuencia de Registro de Ventas')
plt.ylabel('Número de Respuestas')
plt.legend(title='Importancia de Reportes de Ventas')
plt.show()
```
